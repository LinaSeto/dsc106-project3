## Fire Index

```python
def retrieve_xarr(zstore):
    mapper = gcs.get_mapper(zstore) # create a mutable-mapping-style interface to the store
    ds = xr.open_zarr(mapper, consolidated=True) # open it using xarray and zarr
    return ds
```

### Write to JSON file

```python
from dask.diagnostics import ProgressBar

STRIDE      = 2
CACHE_DIR   = 'fire_cache'
EXPERIMENTS = ['ssp126', 'ssp245', 'ssp370', 'ssp585']
INPUT_VARS  = ['tas', 'pr', 'sfcWind', 'hurs']
MEMBER_TRY  = ('r4i1p1f1', 'r1i1p1f1', 'r10i1p1f1', 'r11i1p1f1')

os.makedirs(CACHE_DIR, exist_ok=True)

def find_zstore_idx(experiment, variable):
    for member in MEMBER_TRY:
        mask = (
            (zstores['source_id']     == 'CESM2')      &
            (zstores['experiment_id'] == experiment)   &
            (zstores['variable_id']   == variable)     &
            (zstores['table_id']      == 'day')        &
            (zstores['member_id']     == member)       &
            (zstores['grid_label']    == 'gn')
        )
        matches = zstores[mask]
        if len(matches) > 0:
            return matches.index[0], member
    raise ValueError(
        f"No daily store found for {experiment}/{variable} across members {MEMBER_TRY}"
    )


def load_and_prep(experiment, variable, stride=STRIDE):
    idx, member = find_zstore_idx(experiment, variable)
    ds = retrieve_xarr(zstores.iloc[idx].zstore)
    da = ds[variable]
    da = da.isel(lat=slice(None, None, stride),
                 lon=slice(None, None, stride))
    da = da.chunk({'time': -1, 'lat': 24, 'lon': 36})
    return da, member


def compute_fire_for_ssp(experiment, out_path):
    print(f"{experiment} — loading inputs...", flush=True)
    inputs = {}
    used_members = {}
    for var in INPUT_VARS:
        inputs[var], used_members[var] = load_and_prep(experiment, var)
    if len(set(used_members.values())) > 1:
        print(f"  ⚠ mixed members for {experiment}: {used_members}", flush=True)

    print(f"{experiment} — computing FWI and saving to disk...", flush=True)
    fire = xclim.indices.fire._cffwis.cffwis_indices(
        tas     = inputs['tas'],
        pr      = inputs['pr'],
        sfcWind = inputs['sfcWind'],
        hurs    = inputs['hurs'],
        lat     = inputs['tas'].lat,
    ).FWI
    fire_yearly = fire.groupby('time.year').mean().squeeze()

    with ProgressBar():
        fire_yearly.to_netcdf(out_path)
    print(f"{out_path}\n", flush=True)

    del inputs, fire, fire_yearly

for experiment in EXPERIMENTS:
    out_path = os.path.join(CACHE_DIR, f'fire_{experiment}.nc')
    if os.path.exists(out_path):
        print(f"{experiment} — cached, skipping ({out_path})\n", flush=True)
        continue
    try:
        compute_fire_for_ssp(experiment, out_path)
    except Exception as e:
        print(f"{experiment} failed: {e}\n", flush=True)
        continue 

print("All done. Cached files:")
for f in sorted(os.listdir(CACHE_DIR)):
    size_mb = os.path.getsize(os.path.join(CACHE_DIR, f)) / 1e6
    print(f"  {f}  ({size_mb:.1f} MB)")
```

```python
fire_arrays = {
    ssp: xr.open_dataarray(f'{CACHE_DIR}/fire_{ssp}.nc').load()
    for ssp in EXPERIMENTS
}
```

## Near-Surface Air Temperature, Leaf Area Index, and Sea Surface Height Above Geoid

```python
catalog = pd.read_csv('https://storage.googleapis.com/cmip6/cmip6-zarr-consolidated-stores.csv')

def open_store(zstore):
    return xr.open_zarr(zstore, consolidated=True, storage_options={'token': 'anon'})

def find_store(experiment_id, variable_id,
               member_id='r4i1p1f1', activity_id='ScenarioMIP', table_id='Amon'):
    mask = (
        (catalog['activity_id']   == activity_id)   &
        (catalog['source_id']     == 'CESM2')       &
        (catalog['experiment_id'] == experiment_id) &
        (catalog['variable_id']   == variable_id)   &
        (catalog['table_id']      == table_id)      &
        (catalog['member_id']     == member_id)     &
        (catalog['grid_label']    == 'gn')
    )
    matches = catalog[mask]
    if len(matches) == 0:
        raise ValueError(f"No store: {experiment_id}/{variable_id}/{table_id}/{member_id}")
    return matches.iloc[0]['zstore']


SSPS = ['ssp126', 'ssp245', 'ssp370', 'ssp585']
VAR_CONFIG = {
    'tas': {'table_id': 'Amon', 'transform': lambda x: x - 273.15, 'unit': 'degC'},
    'lai': {'table_id': 'Lmon', 'transform': lambda x: x, 'unit': 'm2_m-2'},
    'zos': {'table_id': 'Omon', 'transform': lambda x: x * 100, 'unit': 'cm'},
}
STRIDE  = 2
OUT_DIR = 'cmip6_json'

def fetch_annual(ssp, var):
    cfg = VAR_CONFIG[var]
    last_err = None
    for member in ('r4i1p1f1', 'r1i1p1f1'):
        try:
            ds = open_store(find_store(ssp, var,
                                       table_id=cfg['table_id'],
                                       member_id=member))
            break
        except ValueError as e:
            last_err = e
    else:
        raise RuntimeError(f"No store for {var}/{ssp}: {last_err}")
    annual = ds[var].groupby('time.year').mean('time')
    return cfg['transform'](annual)

def downsample(da, stride):
    spatial = [d for d in da.dims if d in ('lat', 'lon', 'nlat', 'nlon')]
    return da.isel({d: slice(None, None, stride) for d in spatial})

os.makedirs(OUT_DIR, exist_ok=True)
manifest = {'variables': {}, 'ssps': SSPS, 'stride': STRIDE}

for var, cfg in VAR_CONFIG.items():
    manifest['variables'][var] = {'unit': cfg['unit'], 'files': {}}
    for ssp in SSPS:
        print(f"Processing {var} / {ssp} ...", flush=True)
        annual = downsample(fetch_annual(ssp, var), STRIDE).load()

        years = [int(y) for y in annual.year.values]

        df = annual.to_dataframe(name='value').reset_index()
        pivoted = (df.pivot_table(index=['lat', 'lon'],
                                  columns='year',
                                  values='value',
                                  aggfunc='first')
                     .reindex(columns=years)
                     .dropna(how='all')
                     .round(3))
        cells = [
            {'lat': round(float(lat), 3),
             'lon': round(float(lon), 3),
             'v':   [None if pd.isna(x) else float(x) for x in row.values]}
            for (lat, lon), row in pivoted.iterrows()
        ]

        output = {
            'variable': var,
            'ssp':      ssp,
            'unit':     cfg['unit'],
            'years':    years,
            'cells':    cells,
        }

        out_path = os.path.join(OUT_DIR, f'{var}_{ssp}.json')
        with open(out_path, 'w') as f:
            json.dump(output, f, separators=(',', ':'))

        size_mb = os.path.getsize(out_path) / 1e6
        print(f"  → {out_path}  ({len(cells):,} cells × {len(years)} yrs, {size_mb:.1f} MB)")

        manifest['variables'][var]['files'][ssp] = {
            'path':      f'{var}_{ssp}.json',
            'n_cells':   len(cells),
            'year_range': [years[0], years[-1]],
        }

with open(os.path.join(OUT_DIR, 'manifest.json'), 'w') as f:
    json.dump(manifest, f, indent=2)
print(f"\nManifest: {OUT_DIR}/manifest.json")
```
