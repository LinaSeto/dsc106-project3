# Data Transformations

```python
import pandas as pd
import numpy as np
import xarray as xr
import zarr
import gcsfs
import json
import xclim

gcs = gcsfs.GCSFileSystem(token='anon')
zstores = pd.read_csv('https://storage.googleapis.com/cmip6/cmip6-zarr-consolidated-stores.csv')
```

## Fire Index

```python
def retrieve_xarr(zstore):
    mapper = gcs.get_mapper(zstore) # create a mutable-mapping-style interface to the store
    ds = xr.open_zarr(mapper, consolidated=True) # open it using xarray and zarr
    return ds
```

### SSP1 Scenario

```python
ssp1_tas = retrieve_xarr(zstores.iloc[444426].zstore)
ssp1_pr = retrieve_xarr(zstores.iloc[444581].zstore)
ssp1_sfcWind = retrieve_xarr(zstores.iloc[444424].zstore)
ssp1_hurs = retrieve_xarr(zstores.iloc[444593].zstore)
```

```python
ssp1_fire = xclim.indices.fire._cffwis.cffwis_indices(
    tas=ssp1_tas.tas.chunk(dict(time=-1)), 
    pr=ssp1_pr.pr.chunk(dict(time=-1)), 
    sfcWind=ssp1_sfcWind.sfcWind.chunk(dict(time=-1)), 
    hurs=ssp1_hurs.hurs.chunk(dict(time=-1)), 
    lat=ssp1_tas.lat).FWI
ssp1_fire_yearly = ssp1_fire.groupby('time.year').mean().squeeze()
ssp1_fire_yearly
```
